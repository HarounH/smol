# Text Diffusion Run Analysis

Load one or more training runs from `runs/`, inspect the available artifacts, and plot metrics with Plotly.

The notebook reads:
- `logs/train_metrics.jsonl`
- `logs/events.jsonl`
- `run_summary.json`


```python
from smol.diffusion.text.analysis import (
    configure_plotly_renderer,
    describe_run,
    discover_runs,
    expand_internal_metric_keys,
    find_repo_root,
    plot_metric_grid,
    plot_metric_panel_grid,
    select_runs,
    show_figure,
)

REPO_ROOT = find_repo_root()
RUNS_ROOT = REPO_ROOT / "runs"
PLOTLY_RENDERER = configure_plotly_renderer()
print(f"repo root: {REPO_ROOT}")
print(f"runs root: {RUNS_ROOT}")
print(f"plotly renderer: {PLOTLY_RENDERER}")
```

```python
all_runs = discover_runs(RUNS_ROOT)
print(f"discovered {len(all_runs)} run(s)")
for run in all_runs:
    print(describe_run(run))
```

```python
# Notebook controls.
# Set RUN_STAMPS to specific timestamps to compare exact runs.
# Leave RUN_STAMPS = None to automatically select the latest MAX_AUTO_RUNS.

EXPERIMENT_NAME = "scale-300m-fineweb-1bt"
RUN_STAMPS = ["20260501-080937", "20260503-154905"]
MAX_AUTO_RUNS = 3
SMOOTHING_SIGMA = 2.0

TRAIN_METRICS = [
    "loss",
    "corruption_rate",
    "diffusion_timestep_mean",
]

TIMING_METRICS = [
    "timing/data_loading_s",
    "timing/forward_s",
    "timing/backward_s",
]

MEMORY_METRICS = [
    "memory/gpu_allocated_mb",
    "memory/gpu_reserved_mb",
]

INTERNALS_LAYER_INDEX = 0
INTERNALS_SUMMARIES = ["norm", "abs_mean", "abs_max"]
INTERNALS_GRID_COLUMNS = 3


RESIDUAL_BASES = [
    "residual_in",
    "residual_out",
    "residual_in/grad",
    "residual_out/grad",
]

ACTIVATION_BASES = [
    "attention_out",
    "attention_out/grad",
    "mlp_out",
    "mlp_out/grad",
]

QKV_BASES = [
    "q_proj",
    "k_proj",
    "v_proj",
    "q_weight",
    "k_weight",
    "v_weight",
    "q_weight/grad",
    "k_weight/grad",
    "v_weight/grad",
]
```

```python
selected_runs = select_runs(
    all_runs,
    experiment_name=EXPERIMENT_NAME,
    run_stamps=RUN_STAMPS,
    max_auto_runs=MAX_AUTO_RUNS,
)

if not selected_runs:
    raise ValueError("No runs matched the current selection.")

print(f"selected {len(selected_runs)} run(s)")
for run in selected_runs:
    print(describe_run(run))
```

```python
from smol.diffusion.text.analysis import (
    available_metric_keys,
    metric_series,
)

all_keys = available_metric_keys(selected_runs)
space_keys = ([x for x in all_keys if x.startswith("per_char/") and ("space" in x)])
other_keys = [x for x in all_keys if x.startswith("per_char/") and ("token_" in x) and ("_a" in x)]
print(f"{space_keys=}")
print(f"{other_keys=}")
print(type(selected_runs))
```

```python
metric_series(selected_runs[0], space_keys[0])
```

```python
show_figure(plot_metric_grid(selected_runs, space_keys + other_keys, title="Token Metrics", smoothing_sigma=10.0))
```

```python
show_figure(plot_metric_grid(selected_runs, TRAIN_METRICS, title="Training Metrics", smoothing_sigma=10.0))
```

```python
show_figure(plot_metric_grid(selected_runs, TIMING_METRICS, title="Timing Metrics", smoothing_sigma=SMOOTHING_SIGMA))
```

```python
show_figure(plot_metric_grid(selected_runs, MEMORY_METRICS, title="GPU Memory Metrics", smoothing_sigma=SMOOTHING_SIGMA))
```

```python
RESIDUAL_METRICS = expand_internal_metric_keys(
    selected_runs,
    layer_index=INTERNALS_LAYER_INDEX,
    base_names=RESIDUAL_BASES,
    summaries=INTERNALS_SUMMARIES,
)
if RESIDUAL_METRICS:
    show_figure(
        plot_metric_panel_grid(
            selected_runs,
            RESIDUAL_METRICS,
            title=f"Residual Metrics (layer {INTERNALS_LAYER_INDEX})",
            smoothing_sigma=SMOOTHING_SIGMA,
            columns=INTERNALS_GRID_COLUMNS,
        )
    )
else:
    print(f"No residual metrics found for layer {INTERNALS_LAYER_INDEX}.")
```

```python
ACTIVATION_METRICS = expand_internal_metric_keys(
    selected_runs,
    layer_index=INTERNALS_LAYER_INDEX,
    base_names=ACTIVATION_BASES,
    summaries=INTERNALS_SUMMARIES,
)
if ACTIVATION_METRICS:
    show_figure(
        plot_metric_panel_grid(
            selected_runs,
            ACTIVATION_METRICS,
            title=f"Activation Metrics (layer {INTERNALS_LAYER_INDEX})",
            smoothing_sigma=SMOOTHING_SIGMA,
            columns=INTERNALS_GRID_COLUMNS,
        )
    )
else:
    print(f"No activation metrics found for layer {INTERNALS_LAYER_INDEX}.")
```

```python
QKV_METRICS = expand_internal_metric_keys(
    selected_runs,
    layer_index=INTERNALS_LAYER_INDEX,
    base_names=QKV_BASES,
    summaries=INTERNALS_SUMMARIES,
)
if QKV_METRICS:
    show_figure(
        plot_metric_panel_grid(
            selected_runs,
            QKV_METRICS,
            title=f"QKV Metrics (layer {INTERNALS_LAYER_INDEX})",
            smoothing_sigma=SMOOTHING_SIGMA,
            columns=INTERNALS_GRID_COLUMNS,
        )
    )
else:
    print(f"No qkv metrics found for layer {INTERNALS_LAYER_INDEX}.")
```
